```python
#Step1: Import Libraries
import pandas as pd
import numpy as np
```

```python
print(pd.__version__)
```

```python
#Step2: Load Dataset
df = pd.read_excel("Churn_Unclean_Project.xlsx")
```

```python
#Step3: First look
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.shape
```

```python
#Step4: check missing values
df.isnull().sum()
```

```python
# step5: check duplicate records
df.duplicated().sum()
```

```python
#Remove Duplicates
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
#Step6: Replace Dirty Values
df.replace(['N/A','NULL','',' '],np.nan,inplace = True)
```

```python
df.isnull().sum()
```

```python
# Step7: Remove Extra Spaces
df=df.apply(lambda x:x.str.strip() if x.dtype=="object" else x)
```

```python
# Step8: Proper Case
df["Customer_Name"]=df["Customer_Name"].str.title()
df["State"]=df["State"].str.title()
df["City"]=df["City"].str.title()
df["Subscription_Type"]=df["Subscription_Type"].str.title()
df["Contract_Type"]=df["Contract_Type"].str.title()
df["Churn"]=df["Churn"].str.title()
df["Gender"]=df["Gender"].str.title()
```

```python
df["Subscription_Type"].value_counts()
```

```python
df["Subscription_Type"].unique()
```

```python
# Remove leading/trailing spaces and standardize capitalization

df["Subscription_Type"] = df["Subscription_Type"].str.strip().str.title()
```

```python
df["Subscription_Type"].value_counts()
```

```python
#Step9: Standardize Churn Column
df["Churn"].value_counts()
```

```python
#Step10: Covert Numeric Columns
cols=["Age","Tenure_Months","Monthly_Charges","Total_Charges"]

for c in cols:
    df[c]=pd.to_numeric(df[c],errors="coerce")
```

```python
df.describe()
```

```python
#Step11: Remove Invalid Age
df = df[(df['Age']>=18)&(df["Age"]<=100)]
```

```python
#Step12: Remove Negative Charges
df=df[df["Monthly_Charges"]>=0]
df=df[df["Total_Charges"]>=0]
```

```python
df.shape
```

```python
#Step13 Covert Date 
df["Last_Interaction_Date"]=pd.to_datetime(df["Last_Interaction_Date"],errors='coerce')
```

```python
#Step14: Fill Missing Values
df["Tech_Support"]=df["Tech_Support"].fillna("No")
df["Payment_Method"]=df["Payment_Method"].fillna("Unknown")
df["Monthly_Charges"]=df["Monthly_Charges"].fillna(df['Monthly_Charges'].mean())
df["Tenure_Months"]=df["Tenure_Months"].fillna(df['Tenure_Months'].mean())
df['Age']=df['Age'].fillna(df['Age'].mean())
```

```python
# Step15: Feature Engineering
# Customer Lifetime Value
df['Customer_value']=df["Monthly_Charges"]*df["Tenure_Months"]
```

```python
df.head()
```

```python
# Monthly Income
df["Monthly_Revenue"]=df["Monthly_Charges"]
```

```python
# Tenure Group  '0-12","13-24","25-48","49-72"
bins =[0,12,24,48,72]
label = labels=["0-12 months","13-24 month","25-48 month","49-72 month"]
df['Tenure_Group']=pd.cut(df["Tenure_Months"],bins=bins,labels=labels)
```

```python
df.head()
```

```python
# Senior Citizen Flag

df["Senior_Flag"]=np.where(df["Age"]>=60,"Senior","Adult")
```

```python
df.head()
```

```python
# Churn Flag

df["Churn_Flag"]=df["Churn"].map({"Yes":1,"No":0})
```

```python
df.head()
```

```python
#Step16: Final Check

df.info()
```

```python
df["Last_Interaction_Date"]=df["Last_Interaction_Date"].fillna(df["Last_Interaction_Date"].median())
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
# Step17 CSV Clean Dataset

df.to_csv("Clean_Churn_Data.csv",index=False)
```
